```python
# 2. ORIGINAL DATASET

print("\nORIGINAL DATASET\n")
print("-"*68)
display(df)
```

```python
# 3. FIRST 5 RECORDS

print("\nFIRST 5 RECORDS\n")
print("-"*68)
display(df.head())   # by default head will print first 5, If  want to change specify the number
print("-"*68)
print("\nLAST 5 RECORDS\n")
display(df.tail())     # By default tail will print last 5, If  want to change specify the number
print("-"*68)
print("\nFirst 2 records\n")
display(df.head(2))
print("-"*68)
```

```python
# 4. DATASET SHAPE

print("DATASET SHAPE")
print(df.shape)
```

```python
# 5. COLUMN NAMES

print("COLUMN NAMES")
print(df.columns)
```

```python
# 6. DATA TYPES

print("DATA TYPES")
print(df.dtypes)
```

```python
# 7. DESCRIPTIVE STATISTICS

print("\nDESCRIPTIVE STATISTICS\n")
print("-"*42)
display(df.describe())
print("-"*42)
```

```python
# 8. MISSING VALUES

print("MISSING VALUES")
print(df.isnull().sum())
```

```python
# 9. ROWS CONTAINING MISSING VALUES

print("ROWS CONTAINING MISSING VALUES")
display(df[df.isnull().any(axis=1)])
```

```python
# 10. TOTAL DUPLICATE RECORDS

print("TOTAL DUPLICATE RECORDS")
print(df.duplicated().sum())
```

```python
# 11. DUPLICATE RECORDS

print("DUPLICATE RECORDS")
display(df[df.duplicated()])
```
