```python
# 1. Import NumPy

import numpy as np
```

```python
# 2. Create a NumPy Array

arr = np.array([10 , 20 ,30 ,40 ,50])

print("Numpy Array")
print(arr)
```

```python
# 3. Array Indexing

print("\n--- Array Indexing ---\n")
print("First Element : ",arr[0])
print("3rd Element : ",arr[2])
print("Last Element : ",arr[-1])
```

```python
# 4. Array Slicing

print("\n--- Array Slicing ---\n")
print("First two elements : ",arr[0:2])
print("Element from index 2 :",arr[2:])
print("Last Three Elements : ",arr[-3:])
```

```python
# 5. Basic Array Operations

print("\n--- Basic Array Operations ---\n")
print("Add 5 :",arr+5)
print("Subtract 3 :",arr - 3)
print("Multiply 4 :",arr *4)
print("Divide by 2 :",arr/2)
print("Modulus of 3",arr % 3)
```

```python
# 6. Mathematical Operations

print("\n--- Mathematical Operations ---\n")
print("Sum :",np.sum(arr))
print("Mean :",np.mean(arr))
print("Maximum :",np.max(arr))
print("Minimum :",np.min(arr))
```

```python
# 8. Shape of the original array

print("Shape of the original Array")

print(arr.shape)
```

```python
# 7. Reshaping an Array

print("\n--- Array Reshaping ---")
reshaped_arr = arr.reshape(5,1)

print("Reshaped Array")
print(reshaped_arr)
```

```python
# 8. Import Pandas

import pandas as pd
```

```python
# 9. Create Sample Data

data = {
"Name": ["Asha", "Ravi", "Meera", "Karan"],
"Marks": [88, 72, 91, 65],
"Branch": ["CSE", "ECE", "CSE", "ISE"]
}
```

```python
# 10. Create a DataFrame
df = pd.DataFrame(data)
print("\n--- Pandas DataFrame ---")
display(df)
```

```python
# 11. Column Indexing

print("\n--- Column Indexing ---")
print("Name Column:")
print(df["Name"])
print("\nMarks Column:")
print(df["Marks"])
```

```python
# 12. Row Indexing

print("\n--- Row Indexing ---")
print("First Row:")
print(df.iloc[0])
print("\nThird Row:")
print(df.iloc[2])
```

```python
# 13. Access a Specific Value

print("\n--- Specific Value ---")
print("Marks of first student:", df.iloc[0, 1])
```

```python
# 14. DataFrame Row Slicing

print("\n--- Row Slicing ---")
print("First two rows:")
print(df.iloc[0:2])
```

```python
# 15. Row and Column Slicing

print("\n--- Row and Column Slicing ---")
print("First three rows with Name and Marks:")
print(df.iloc[0:3, [0, 1]])
```

```python
# 16. Filtering Data

print("\n--- Data Filtering ---")
result = df[df["Marks"] > 70]
print("Students with Marks greater than 70:")
print(result)
```
