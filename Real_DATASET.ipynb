```python
import pandas as pd
```

```python
# 1. Load the dataset

df = pd.read_csv("indian_stock_market.csv")
```

```python
# 2. Display the original dataset

print("ORIGINAL DATASET")
display(df)
```

```python
# 3. Display the first 5 records

print("FIRST 5 RECORDS")
display(df.head())
```

```python
# 4. Display the last 5 records

print("LAST 5 RECORDS")
display(df.tail())
```

```python
# 5. Display the number of rows and columns

print("DATASET SHAPE")
print(df.shape)
```

```python
# 6. Display column names

print("COLUMN NAMES")
print(df.columns)
```

```python
# 7. Display data types

print("DATA TYPES")
print(df.dtypes)
```

```python
# 8. TOTAL DUPLICATE RECORDS

print("TOTAL DUPLICATE RECORDS")
print(df.duplicated().sum())
```

```python
# 9. MISSING VALUES

print("MISSING VALUES")
print("-"*40)
print(df.isnull().sum())
```

```python
# 10. ROWS CONTAINING MISSING VALUES

print("ROWS CONTAINING MISSING VALUES")
display(df[df.isnull().any(axis=1)])
```
