```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("students.csv")
```

```python
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Marks"], bins=10, edgecolor="black")

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Histogram of Marks")

plt.show()
```

```python
df[["StudyHours", "Attendance", "PrevMarks", "Marks"]].hist(
    figsize=(12,8),
    bins=10,
    edgecolor="black"
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))

pass_data = df[df["Result"] == "Pass"]["Marks"]
fail_data = df[df["Result"] == "Fail"]["Marks"]

box = plt.boxplot(
    [fail_data, pass_data],
    vert=False,
    patch_artist=True,
    tick_labels=["Fail", "Pass"]
)

# Colors
box["boxes"][0].set_facecolor("red")
box["boxes"][1].set_facecolor("green")

plt.xlabel("Marks")
plt.title("Marks Distribution: Pass vs Fail")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df[["StudyHours", "Attendance", "PrevMarks", "Marks"]]
)

plt.title("Box Plot of Numerical Variables")
plt.ylabel("Values")

plt.show()
```

```python
plt.figure(figsize=(8,5))

# Pass = Green
pass_data = df[df["Result"] == "Pass"]
plt.scatter(
    pass_data["StudyHours"],
    pass_data["Marks"],
    color="green",
    label="Pass"
)

# Fail = Red
fail_data = df[df["Result"] == "Fail"]
plt.scatter(
    fail_data["StudyHours"],
    fail_data["Marks"],
    color="red",
    label="Fail"
)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,5))

pass_data = df[df["Result"] == "Pass"]
fail_data = df[df["Result"] == "Fail"]

plt.scatter(
    pass_data["StudyHours"],
    pass_data["Marks"],
    color="green",
    label="Pass"
)

plt.scatter(
    fail_data["StudyHours"],
    fail_data["Marks"],
    color="red",
    label="Fail"
)

# Best-fit line
import numpy as np

x = df["StudyHours"]
y = df["Marks"]

m, c = np.polyfit(x, y, 1)
plt.plot(x, m*x + c, color="black", label="Best-fit line")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks with Best-Fit Line")
plt.legend()

plt.show()
```

```python
average_marks = df.groupby("Branch")["Marks"].mean()

plt.figure(figsize=(8,5))

plt.bar(average_marks.index, average_marks.values)

plt.xlabel("Branch")
plt.ylabel("Average Marks")
plt.title("Average Marks by Branch")

plt.show()
```

```python
result_count = df["Result"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    result_count.values,
    labels=result_count.index,
    autopct="%1.1f%%"
)

plt.title("Pass vs Fail Distribution")

plt.show()
```

```python
sns.pairplot(
    df[
        ["StudyHours", "Attendance", "PrevMarks", "Marks"]
    ]
)

plt.show()
```

```python
plt.figure(figsize=(9,7))

correlation = df[
    ["StudyHours", "Attendance", "PrevMarks", "Marks"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
