```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4, 5])
print(a)
```

```python
print(a.ndim)    # Number of dimensions
print(a.shape)   # Shape
print(a.size)    # Number of elements
print(a.dtype)   # Data type
```

```python
b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(b)
print(b.shape)
print(b.ndim)
```
